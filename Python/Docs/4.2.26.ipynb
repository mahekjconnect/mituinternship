# FEATURE ENCODING

- Feature Encoding means converting categorical data into numerical form
- Machine Learning models cannot understand text or strings
- Models only perform mathematical operations
- Therefore categorical values must be converted into numbers

- There are two types of categorical data:
    - Nominal data
    - Ordinal data


# NOMINAL DATA

- Nominal data has no order or ranking
    - Example:
        - Red, Blue, Green
        - Male, Female
        - Mumbai, Delhi, Pune
- No category is greater or smaller than another
- Order does not matter


# ORDINAL DATA

- Ordinal data has a clear order or ranking
- Example:
    - Poor < Average < Good
    - Small < Medium < Large
    - School < UG < PG
- Here order matters
- Higher category represents higher level


# ORDINAL ENCODING

- Used when categorical feature has order
- Converts categories into numbers based on ranking
- Example:
    - Poor → 0
    - Average → 1
    - Good → 2
- The numerical values preserve the order
- Model understands that 2 > 1 > 0
- Should only be used when real order exists
- If used on nominal data, it creates fake relationships


# LABEL ENCODING

- Usually used for target variable (y)
- Converts class labels into numbers
- Example:
    - No → 0
    - Yes → 1
    - Cat → 0
    - Dog → 1
    - Tiger → 2
- Mostly used for output column
- Not recommended for input features because it may create false ordering


# DIFFERENCE BETWEEN ORDINAL ENCODING AND LABEL ENCODING

- OrdinalEncoder:
    - Used for input features (X)
    - Can handle multiple columns
    - Used when order exists

- LabelEncoder:
    - Used for target column (y)
    - Works on single column
    - Automatically assigns numbers based on alphabetical order


# IMPORTANT MACHINE LEARNING RULE

- Always fit encoder on training data only
- Never fit encoder on test data
- Test data should only be transformed
- This prevents data leakage


# WHY ENCODING IS IMPORTANT

- Models perform mathematical operations
- Strings cannot be used in mathematical equations
- Encoding makes categorical data usable for ML algorithms

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(df1.iloc[:,:2], df1.iloc[:,-1], test_size=0.3, random_state=0)

X_train.shape, X_test.shape

```

```python
from sklearn.preprocessing import OrdinalEncoder
oe = OrdinalEncoder(categories = [['Poor', 'Average', 'Good'],['School','UG','PG']])
oe.fit(X_train)
X_train= oe.transform(X_train)
X_train
```

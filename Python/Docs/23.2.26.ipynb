# Decision Tree

- Decision Tree is a supervised machine learning algorithm
- It works like a flowchart (question-answer structure)
- It splits data into smaller parts based on conditions
- It can be used for both Classification and Regression

# Structure of Decision Tree

- Root Node → First question / first split
- Decision Node → Further questions
- Leaf Node → Final output / prediction
- Branch → Result of a condition (Yes/No, True/False)

# How Decision Tree Works

- The model selects the best feature to split the data
- It divides the dataset into smaller subsets
- It keeps splitting until stopping condition is met
- Final nodes give prediction

# Decision Tree for Classification

- Used when output is categorical (Yes/No, Pass/Fail, Spam/Not Spam)
- It tries to make groups as pure as possible
- Purity means most values in a group belong to one class

- Metrics used:
  - Gini Impurity
  - Entropy
  - Information Gain

# Decision Tree for Regression

- Used when output is numerical (Salary, Price, Marks)
- It splits data to reduce error
- Metric used:
  - Mean Squared Error (MSE)

# Important Concepts

- Splitting → Dividing data based on a condition
- Depth → Number of levels in the tree
- Parent Node → Node that splits
- Child Node → Result of split

# Overfitting in Decision Tree

- Happens when tree becomes too deep
- Model memorizes training data
- Performs poorly on new data

# How to Prevent Overfitting

- Limit max_depth
- Increase min_samples_split
- Increase min_samples_leaf
- Use Pruning

# Important Hyperparameters (sklearn)

- max_depth → Maximum depth of tree
- min_samples_split → Minimum samples required to split a node
- min_samples_leaf → Minimum samples required at a leaf node
- criterion → Metric to measure split (gini or entropy)

# Advantages

- Easy to understand and interpret
- No need for feature scaling
- Works with both numerical and categorical data

# Disadvantages

- Can overfit easily
- Sensitive to small changes in data
- Large trees become complex

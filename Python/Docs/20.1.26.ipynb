# Seaborn 


seaborn is built on top of matplotlib thats why we need to import both of these libraries.

```python
import seaborn as sns
import matplotlib.pyplot as plt 


data = sns.load_dataset('tips')
data.head()
```

in this scatterplot we are trying to understand the relation between the 'total_bill' and 'tip'. we also added 'smokers' category to understand better relations based on category too.

```python
sns.scatterplot(
    data =data,
    x = 'total_bill',
    y = 'tip',
    hue = 'smoker' )
plt.show()
```

regplot = regression plot, shows the data(dots), and draws a best-fitting line
- matlab ki the best expected outcome ya fir probabality ye hai- aesa batara hai vo line. 
- see in this plot it is showing that the more the bill the higher the tip will be, so the line is basically a excepted outcome of the tips. 
- regplot basically just shows data and its expected or more probabal outcome.
- used when X and Y are numerical and when u want to see the trend
- ab isme jitne vo side me faded shade ki hui jagha hai na that is confidence band, wider band = more uncertaninty, narrow band = less uncertanity.

```python
sns.regplot(data=data, x="total_bill", y="tip")
plt.show()
```

lmplot = linear model plot
- basically grouped regression plot, more than 1 line, gives different expected outcome lines for different categories
- so u can compare the expected outcome of different categories in the same plot 
- used when relationship might differ by category


```python
sns.lmplot(data=data, x="total_bill", y="tip", hue="smoker")
plt.show()
```

boxplot - it shows distribution not relationships
- median(the middle value, 50% above and 50% below), speard(shows if the values are tightly packed or all over the place), and outliners(unsual or unexpected values)
- used when we are comparies a numeric column across categories.

```python
sns.boxplot(data=data, x="smoker", y="tip")
plt.show()
```

```python
sns.boxplot(data=data, x="day", y="total_bill")
plt.show()
```

```python
sns.pairplot(
    data = data,
    vars = ['total_bill', 'tip', 'size'],
    hue = 'smoker'
)
plt.show()
```

# Kernels in Support Vector Machine (SVM)

- Kernel decides what type of boundary SVM will draw to separate the data

- Sometimes a straight line cannot separate the data
- So kernel helps create different shapes of boundaries

# Linear Kernel

- Draws a straight line between classes
- Used when data can be separated easily

Example idea:
🔵 🔵 🔵 | 🔴 🔴 🔴

kernel = "linear"

# Polynomial Kernel

- Creates a curved boundary
- Used when data has a curved pattern
- More flexible than linear

Example idea:
🔵 🔵 🔵
   🔴 🔴
🔵 🔵 🔵

kernel = "poly"

# RBF Kernel (Radial Basis Function)

- Most commonly used kernel
- Very flexible boundary
- Can create circular or complex shapes
- Works well for complicated datasets

Example idea:
🔵 🔵 🔵 🔵
🔵 🔴 🔴 🔵
🔵 🔴 🔴 🔵
🔵 🔵 🔵 🔵

kernel = "rbf"

# Sigmoid Kernel

- Creates S-shaped boundary
- Similar to activation function in neural networks
- Not used very often

kernel = "sigmoid"

# Quick Memory Trick

Linear → Straight line  
Polynomial → Curved boundary  
RBF → Very flexible / complex shapes  
Sigmoid → S-shaped boundary

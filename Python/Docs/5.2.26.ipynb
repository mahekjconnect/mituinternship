# ONE HOT ENCODING (OHE)

- Machine learning models cannot understand categorical text data.
- Example: Education = ["Bachelors", "Masters", "PhD"]

- If we encode like:
  - Bachelors = 0
  - Masters   = 1
  - PhD       = 2

- The model assumes:
  - PhD > Masters > Bachelors
- This creates a FALSE ordering relationship.
- These are categories, not numerical quantities.


# Solution: One Hot Encoding

- Create separate binary columns for each category.

- Example:

  Education   Bachelors   Masters   PhD
  --------------------------------------
  Bachelors       1          0       0
  Masters         0          1       0
  PhD             0          0       1

- Each row has:
  - One "1"
  - Rest "0"

- This is called One Hot Encoding.


# Pandas Implementation

- pd.get_dummies(df["Education"])


# DUMMY VARIABLE TRAP (MULTICOLLINEARITY)

- After One Hot Encoding:

  Bachelors   Masters   PhD
  ----------------------------
      1          0        0
      0          1        0
      0          0        1

- Notice:
  - For every row:
    - Bachelors + Masters + PhD = 1

- That means:
  - If we know two columns, we can calculate the third.
  - Example:
    - PhD = 1 - (Bachelors + Masters)

- This is called PERFECT MULTICOLLINEARITY.


# What is Multicollinearity?

- Multicollinearity = When one feature can be predicted using other features.
- It mainly affects:
  - Linear Regression
  - Logistic Regression

- Problems it causes:
  - Coefficients become unstable
  - Model cannot decide which feature is important
  - Matrix inversion becomes problematic


# Dummy Variable Trap

- When we create ALL dummy columns,
  - One column becomes redundant.
- This situation is called the Dummy Variable Trap.


# How to Fix Dummy Variable Trap

- Drop one dummy column.
- The dropped column becomes the reference category.

- Example after dropping one column:

  Bachelors   Masters
  ----------------------
      1          0
      0          1
      0          0   → This means PhD

- Now:
  - No perfect multicollinearity exists.


# Python Solution

- Using pandas:
  - pd.get_dummies(df["Education"], drop_first=True)

- Using sklearn:
  - OneHotEncoder(drop="first")


# Important Notes

- Drop one column if using:
  - Linear Regression
  - Logistic Regression

- Not necessary for:
  - Decision Trees
  - Random Forest
  - XGBoost

- Tree-based models are not affected by multicollinearity.

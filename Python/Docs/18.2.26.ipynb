# Regularization

- Regularization is used to reduce overfitting
- It makes the model simpler
- It prevents the model from memorizing the data
- It controls how big the coefficients can become


# Why Do We Need Regularization?

- Sometimes the model becomes too complex
- It fits training data very well
- But performs badly on new (test) data
- This is called overfitting
- Regularization helps fix this problem


# Basic Idea

- Normal Linear Regression tries to reduce error
- Regularization reduces:
  - Error
  - And keeps coefficients small
- Smaller coefficients → simpler model


# What Does It Actually Do?

- Adds a penalty if coefficients become too large
- Large coefficients = complex model
- Regularization pushes coefficients toward smaller values


# Types of Regularization

- Ridge Regression
- Lasso Regression
- Elastic Net


# Ridge Regression (L2)

- Shrinks coefficients
- Makes them smaller
- But does NOT make them zero
- Good when all features are important


# Lasso Regression (L1)

- Shrinks coefficients
- Can make some coefficients exactly zero
- Automatically removes unimportant features
- Works like feature selection


# Elastic Net

- Combination of Ridge and Lasso
- Shrinks coefficients
- Can also remove some features
- Useful when features are correlated


# What is Lambda (λ)?

- Lambda controls how strong the penalty is
- Small λ → little penalty → model behaves normally
- Large λ → strong penalty → model becomes simpler
- Very large λ → model may underfit


# Effect of Regularization

- Reduces overfitting
- Makes model more stable
- Reduces variance
- Slightly increases bias


# Simple Understanding

- Overfitting = model too smart
- Underfitting = model too simple
- Regularization = keeps model balanced


# Key Idea

- It controls model complexity
- Helps model perform better on unseen data

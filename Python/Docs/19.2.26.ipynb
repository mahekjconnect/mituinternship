# Classification

- Classification is a type of supervised machine learning
- It is used to predict categories or classes
- Output is not a number
- Output is a label


# What is Classification Used For?

- Spam or Not Spam
- Disease or No Disease
- Fraud or Not Fraud
- Pass or Fail
- Cat or Dog


# Difference Between Regression and Classification

- Regression → predicts continuous numbers
- Classification → predicts categories


# Types of Classification

- Binary Classification
  - Only two classes
  - Example: Yes / No

- Multi-class Classification
  - More than two classes
  - Example: Red / Blue / Green

- Multi-label Classification
  - One input can have multiple labels
  - Example: Movie can be Action + Comedy


# How Classification Works

- Model learns patterns from training data
- Finds decision boundary
- Uses that boundary to separate classes
- Predicts class for new data


# What is a Decision Boundary?

- A line (or curve) that separates classes
- In 2D → line
- In 3D → plane
- In higher dimensions → hyperplane


# Common Classification Algorithms

- Logistic Regression
- K-Nearest Neighbors (KNN)
- Decision Tree
- Random Forest
- Support Vector Machine (SVM)
- Naive Bayes
- Neural Networks


# Evaluation Metrics for Classification

- Accuracy
- Precision
- Recall
- F1 Score
- Confusion Matrix
- ROC-AUC


# What is Accuracy?

- Accuracy = Correct Predictions / Total Predictions
- Works well when classes are balanced
- Not good for imbalanced datasets


# Key Idea

- Classification predicts labels
- It separates data into categories
- Used when output is discrete, not continuous

# Accuracy Metrics for Regression

- Regression is used to predict continuous numerical values
- We cannot use classification accuracy (like % correct)
- Instead, we measure how close predictions are to actual values
- These are called regression evaluation metrics


# Mean Absolute Error (MAE)

- Formula: MAE = average(|Actual - Predicted|)
- It calculates average absolute difference
- All errors are treated equally
- Easy to understand
- Unit is same as target variable
- Less sensitive to outliers compared to MSE


# Mean Squared Error (MSE)

- Formula: MSE = average((Actual - Predicted)^2)
- Errors are squared before averaging
- Penalizes large errors more
- Sensitive to outliers
- Unit becomes squared (not same as original unit)


# Root Mean Squared Error (RMSE)

- Formula: RMSE = sqrt(MSE)
- Square root of MSE
- Unit becomes same as target variable
- Penalizes large errors
- Most commonly used regression metric


# R-squared (R² Score)

- Measures how much variance in target is explained by model
- Formula: R² = 1 - (SS_res / SS_total)
- Value range: (-∞ to 1)
- R² = 1 → perfect model
- R² = 0 → model explains nothing
- R² < 0 → model is worse than predicting mean


# Adjusted R-squared

- Modified version of R²
- Penalizes model for adding unnecessary features
- Useful in multiple linear regression
- Prevents overfitting due to too many variables


# Mean Absolute Percentage Error (MAPE)

- Formula: average(|(Actual - Predicted) / Actual|) × 100
- Expressed in percentage
- Easy to interpret
- Problem: does not work well when actual value is zero


# When to Use Which Metric

- MAE → when you want simple average error
- MSE → when large errors should be heavily penalized
- RMSE → when you want error in original units
- R² → when you want to know model explanatory power
- Adjusted R² → when using multiple features
- MAPE → when percentage error matters (business problems)


# Quick Comparison

- MAE → robust, simple
- MSE → sensitive to outliers
- RMSE → most popular
- R² → goodness of fit measure
- Adjusted R² → feature-aware R²
- MAPE → percentage-based error

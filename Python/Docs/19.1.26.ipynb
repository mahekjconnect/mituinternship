# Matplotlib in Python 

Matplotlib is widely used python library for static, interactive, and animated visualization from data.

matplotlib helps u - visualize data, understand trends, compare values, make charts for reports/ ML / data science.

```python
import matplotlib.pyplot as plt 

x= [0,1,2,3,4]
y=[0,1,4,9,16]

plt.plot(x,y)
plt.show()
```

Figure (paper)
 └── Axes (graph area)
      ├── x-axis
      └── y-axis


```python
#line plot
x= [0,2,4,6,8]
y=[0,1,4,9,16]
plt.plot(x,y, marker='o', linestyle='-', color='b') #customaized line plot
plt.title("Line Plot Example") #title
plt.xlabel("X-axis") #custom x-axis label
plt.ylabel("Y-axis") #custom y-axis label
plt.show() #display the plot
```

line plots (trends over time) are usally used to show
- temprature
- stock prices
- growth

```python
#bar chart 
categories = ['A', 'B', 'C', 'D']
values = [5, 7, 3, 4]
plt.bar(categories,values)
plt.title("Bar Chart Example")
plt.xlabel("Categories")
plt.ylabel("Values")
plt.show()
```

Bar chart (comparision) is used to show:
- Student Makrs
- sales per person

```python
#scatter plot
plt.scatter(x,y, color='r', marker='x') #customized scatter plot
plt.title("Scatter Plot Example") #title
plt.xlabel("X-axis") #custom x-axis label
plt.ylabel("Y-axis") #custom y-axis label
plt.grid (True) #enable grid
plt.show() #display the plot
```

```python
a = [2,4,6,8,10]
b = [9,7,5,3,1]
c =[10,8,6,4,2]
plt.plot(a,c,label='Line 1', color='g', linestyle='--', marker='s')
plt.plot(a,b,label='Line 2', color='r', linestyle='-', marker='o')
plt.title("Multiple Lines Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.legend() #legend is used to differentiate lines
plt.show()
```

```python
plt.plot(a,b, color='b',linestyle='--', marker='o', mfc='r', label='2020')
plt.plot(a,c, color='g', linestyle='--', marker='s', mfc='y', label='2021')
plt.title("Multiple Lines with Legend Example")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(16,9)) #custom figure size
plt.subplot(2,1,1) #first subplot
plt.title("First Subplot")
plt.xlabel("X values")
plt.ylabel("Y values")
plt.plot(a,b, color='m', marker='o', linestyle='--')
plt.grid(True)
plt.subplot(2,1,2) #second subplot
plt.title("Second Subplot")
plt.xlabel("X values")
plt.ylabel("Y values")  
plt.plot(a,c, color='c', marker='s', linestyle='-.')
plt.grid(True)
```

```python
courses = ['Math', 'Science', 'History', 'Art']
students = [30, 25, 20, 15]
plt.title("Students Enrolled in Courses")
plt.xlabel("Courses")
plt.ylabel("Number of Students")
plt.bar(courses, students, color='black' , width=0.5, edgecolor='pink')
plt.barh(courses, students, color='blue' ,  edgecolor='pink')
plt.grid(True)
plt.show()
```

```python
#histogram is used to show the distribution of numerical data in bins, bins are intervals meaning range of values
data = np.random.randn(1000)
plt.hist(data, bins=30, color='cyan', edgecolor='black')
plt.title("Histogram Example")
plt.grid(True)
plt.show()
```

# Multiple Linear Regression

- Multiple Linear Regression is an extension of simple linear regression
- It uses more than one input feature to predict a target
- Used when output depends on multiple factors


# Why Do We Need Multiple Regression?

- Real-world problems usually depend on many variables
- Example: House price depends on
  - Area
  - Number of bedrooms
  - Location
  - Age of house
- One feature is not enough


# Equation of Multiple Linear Regression

- y = β0 + β1x1 + β2x2 + β3x3 + ... + βnxn
- y → predicted value
- β0 → intercept
- β1, β2, ... → coefficients
- x1, x2, ... → input features


# Interpretation of Coefficients

- β1 tells how much y changes when x1 increases by 1
- Other variables are assumed constant
- Positive coefficient → direct relationship
- Negative coefficient → inverse relationship


# Assumptions of Multiple Linear Regression

- Linearity → relationship between features and target is linear
- No multicollinearity → features should not be highly correlated
- Homoscedasticity → constant error variance
- Errors should be independent
- Errors should be normally distributed (mainly for inference)


# Multicollinearity Problem

- Happens when features are highly correlated
- Makes coefficient estimates unstable
- Hard to interpret importance of variables
- Can be detected using correlation matrix or VIF


# How Model Learns

- Model minimizes Mean Squared Error (MSE)
- Finds best coefficients β that reduce prediction error
- Uses Normal Equation or Gradient Descent


# Advantages

- More realistic than simple regression
- Can handle multiple influencing factors
- Coefficients are interpretable
- Good baseline model


# Disadvantages

- Sensitive to multicollinearity
- Sensitive to outliers
- Cannot model nonlinear relationships
- Assumptions must hold


# Example Use Cases

- Predict house prices
- Predict salary based on experience and education
- Predict sales using marketing spend, season, price
- Demand forecasting


# Key Idea

- One input → Simple Linear Regression
- Multiple inputs → Multiple Linear Regression
- Still a straight-line relationship in higher dimensions

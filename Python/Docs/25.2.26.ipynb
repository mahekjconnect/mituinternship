# K-Nearest Neighbour (KNN)

- KNN is a supervised machine learning algorithm
- It is used for both classification and regression
- It classifies a new data point based on the nearest data points

# Meaning of K

- K represents the number of nearest neighbours considered
- Example:
  - K = 3 → Look at 3 nearest data points
  - K = 5 → Look at 5 nearest data points

# How KNN Works

- Choose the value of K
- Calculate distance between new point and all training points
- Select K nearest points
- For classification → choose the majority class
- For regression → take the average value

# Distance Metrics Used

- Euclidean Distance (most common)
- Manhattan Distance
- Minkowski Distance

# Advantages

- Simple and easy to understand
- No training phase (lazy learning)
- Works well with small datasets

# Disadvantages

- Slow for large datasets
- Sensitive to noise
- Requires feature scaling

# Why Feature Scaling is Needed

- KNN uses distance calculations
- Large values can dominate the distance
- So features should be scaled (Standardization or Normalization)

# Choosing the Best K

- Small K → Can cause overfitting
- Large K → Can cause underfitting
- Usually choose odd values like 3, 5, 7

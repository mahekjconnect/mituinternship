# Feature Selection

- Process of selecting important features (columns) from dataset
- Removes unnecessary, irrelevant, and redundant features

# Need for Feature Selection

- Improves model accuracy
- Reduces overfitting
- Faster training
- Reduces noise in data
- Makes model simpler and more interpretable

# Types of Feature Selection Techniques

- Filter Methods
- Wrapper Methods
- Embedded Methods


# 1. Filter Methods

- Select features based on statistical measures
- Does NOT use any ML model
- Checks only relevance of features
- Fast and simple

## Techniques

- Information Gain (IG)
  - Measures how much a feature helps in predicting target
  - Higher value = more important feature

- Chi-Square Test (χ²)
  - Used for categorical data
  - Checks dependency between feature and target
  - More dependency = more useful feature

- Correlation Coefficient
  - Measures linear relationship between variables
  - Values range from -1 to +1
  - Remove highly correlated features (to avoid redundancy)


# 2. Wrapper Methods

- Uses ML model to evaluate features
- Trains model multiple times
- Checks usefulness of features
- More accurate but slow

## Techniques

- RFE (Recursive Feature Elimination)
  - Starts with all features
  - Removes least important feature step-by-step
  - Selects best subset of features

- Genetic Algorithm
  - Inspired by natural selection
  - Creates different combinations of features
  - Selects best combination based on model performance


# 3. Embedded Methods

- Feature selection happens during model training
- Combines advantages of filter and wrapper
- Checks both relevance and usefulness

## Techniques

- Decision Trees
  - Automatically select important features
  - Based on metrics like Information Gain or Gini Index
  - Important features appear at top of tree


# Final Comparison

- Filter → Fast, checks relevance only
- Wrapper → Slow, checks usefulness using model
- Embedded → Medium speed, checks both

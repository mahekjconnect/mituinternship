# Logistic Regression

- Logistic Regression is used for classification problems
- It is mainly used for binary classification
- Output is 0 or 1
- Example:
  - Spam / Not Spam
  - Disease / No Disease
  - Pass / Fail


# Why Not Use Linear Regression for Classification?

- Linear regression gives continuous output
- Classification needs output between 0 and 1
- Probabilities must be between 0 and 1
- Linear regression can give values less than 0 or greater than 1


# How Logistic Regression Works

- First, it calculates a linear equation:
  z = β0 + β1x1 + β2x2 + ...

- Then it applies a special function called Sigmoid
- Sigmoid converts any number into range (0,1)


# Sigmoid Function

- Formula:
  1 / (1 + e^(-z))

- If z is very large positive → output close to 1
- If z is very large negative → output close to 0
- If z is 0 → output is 0.5

- It produces an S-shaped curve


# Output of Logistic Regression

- Output is probability
- If probability ≥ 0.5 → class 1
- If probability < 0.5 → class 0
- 0.5 is called threshold (can be changed)


# What is Decision Boundary?

- It is the boundary that separates classes
- Where probability = 0.5
- In 2D → straight line
- In higher dimensions → hyperplane


# Cost Function in Logistic Regression

- We cannot use MSE
- We use Log Loss (Binary Cross Entropy)

- It penalizes wrong predictions heavily
- Helps improve probability estimation


# Assumptions of Logistic Regression

- Relationship between features and log-odds is linear
- No strong multicollinearity
- Large sample size is preferred


# Advantages

- Simple and fast
- Outputs probabilities
- Easy to interpret
- Works well for linearly separable data


# Disadvantages

- Cannot handle complex nonlinear relationships
- Sensitive to outliers
- Needs feature scaling sometimes


# Key Idea

- Logistic Regression is for classification
- It uses linear equation + sigmoid function
- It predicts probability, not direct class

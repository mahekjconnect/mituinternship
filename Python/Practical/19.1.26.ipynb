1.Import matplotlib.pyplot and create a basic line plot showing fare values of the first 10 passengers.

```python
import matplotlib.pyplot as plt
import pandas as pd

df = pd.read_csv("train.csv")
F10 = df.head(10)

```

```python
plt.plot(F10["PassengerId"], F10["Fare"], marker='^', color ='yellow')
plt.xlabel("PassengerId")
plt.ylabel("Fare")  
plt.title("Fare vs PassengerId")
plt.show()
```

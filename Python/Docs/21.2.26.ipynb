# Confusion Matrix

- Confusion Matrix is used to evaluate classification models
- It shows how many predictions were correct and incorrect
- Mainly used in binary classification

- It has 4 parts:
  - True Positive (TP)
  - True Negative (TN)
  - False Positive (FP)
  - False Negative (FN)


# Understanding the 4 Terms

- True Positive (TP)
  - Model predicted Positive
  - And it was actually Positive

- True Negative (TN)
  - Model predicted Negative
  - And it was actually Negative

- False Positive (FP)
  - Model predicted Positive
  - But it was actually Negative
  - Also called Type 1 Error

- False Negative (FN)
  - Model predicted Negative
  - But it was actually Positive
  - Also called Type 2 Error


# Accuracy Score

- Accuracy = (TP + TN) / Total Predictions
- It tells how many predictions were correct
- Works well when classes are balanced
- Not good for imbalanced datasets


# Precision

- Precision = TP / (TP + FP)
- Out of all predicted positives, how many were correct
- Focuses on False Positives
- High precision → fewer false alarms

- Important when:
  - False Positive is costly
  - Example: Spam detection


# Recall (Sensitivity)

- Recall = TP / (TP + FN)
- Out of all actual positives, how many were correctly predicted
- Focuses on False Negatives
- High recall → fewer missed cases

- Important when:
  - Missing a positive case is dangerous
  - Example: Disease detection


# F1 Score

- F1 Score = 2 × (Precision × Recall) / (Precision + Recall)
- It is the harmonic mean of precision and recall
- Useful when classes are imbalanced
- Gives balance between precision and recall


# When to Use What

- Use Accuracy → when dataset is balanced
- Use Precision → when false positives are costly
- Use Recall → when false negatives are costly
- Use F1 Score → when you need balance between precision and recall


# Key Idea

- Accuracy shows overall correctness
- Precision shows quality of positive predictions
- Recall shows how many positives were captured
- F1 balances precision and recall
- Confusion Matrix gives complete picture

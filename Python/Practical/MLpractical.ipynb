```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df =pd.read_csv('train.csv')
df
```

```python
df.drop(['Cabin'], axis=1, inplace=True) #in drop function u need to give a list of columns that why the [], axis =1 means we are dropping a column, inplace = True means we are making the change in the original dataframe
```

```python
df
```

```python
age= df['Age'].median() #median is used to fill the missing values because it is not affected by outliers, we dont use mean because it is affected by outliers, if there are outliers in the Age column then the mean will be skewed and it will not represent the central tendency of the data, median is a better measure of central tendency in this case
print(age)
```

```python
df['Age'].fillna(age, inplace=True) #fillna is used to fill the missing values, inplace = True means we are making the change in the original dataframe
df['Age'].isnull().sum() #to check if there are any missing values in the Age column
```

```python
df
```

```python
df['Embarked'].isnull().sum()
```

```python
embark= df['Embarked'].mode() #mode is used to fill the missing values in categorical columns because it is the most frequent value in the column
print(embark)
```

```python
df['Embarked'].fillna(embark[0],inplace= True) #embark[0] is used to get the first value of the mode because mode can return multiple values if there are multiple modes
df['Embarked'].isnull().sum() #to check if there are any missing values in the Embarked column

```

```python
df
```

```python
#label encoding for 'Sex' because it has only two categories, we can use label encoding to convert it into numerical values
from sklearn.preprocessing import LabelEncoder
le = LabelEncoder() #LabelEncoder is used to convert categorical variables into numerical variables
df['Sex'] = le.fit_transform(df['Sex'])
df['Sex']
```

```python
le.classes_ #to check the classes that are present in the Sex column, it will return an array of classes that are present
```

```python
#one hot encoding for 'Embarked' beacause  it has more than two categories, we can use one hot encoding to convert it into numerical values, also we dont want to give any ordinal relationship to the categories in the Embarked column
df =pd.get_dummies(df, columns=['Embarked'], drop_first=True) #drop_first = True is used to drop the first category because it is redundant, it will not add any new information to the model
df ####NOTEE: encoding should be done in one go only AND RUN IT ONCE ONLYY, otherwise later u cant change thingss......i mean u change but that just gets messed up tbh :)
```

This section of code is just for refrence of how feature scaling is done. Below we will do the train test split first and then scale, to follow the proper pipeline to make a modle :)))))))))

```python
#from sklearn.preprocessing import StandardScaler
#sc = StandardScaler() #standardization means we are scaling the features to have a mean of 0 and a standard deviation of 1, it is used to bring all the features to the same scale, it is important for algorithms that are sensitive to the scale of the features like SVM, KNN, etc.
#df[['Age','Fare']] = sc.fit_transform(df[['Age','Fare']]) #we are scaling the Age and Fare columns because they are numerical columns and they have different scales, we dont want
#df #to check if the features are scaled or not, it should have a mean of 0 and a standard deviation of 1
```

```python
#dropping the columns that are not required for the model
df.drop(['PassengerId','Name','Ticket'], axis=1, inplace=True) #we are dropping the PassengerID, Name and Ticket columns because they are not required for the model, they do not have any predictive power, they are just identifiers for the passengers
df
```

```python
from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test= train_test_split(df.drop(['Survived'],axis=1), df['Survived'],
                                                   test_size=0.3,
                                                   random_state=42,#random_state is used to get the same random numbers every time we run the code, it is used for reproducibility, test_size = 0.3 means we are using 30% of the data for testing and 70% for training
                                                   stratify= df['Survived']) #stratify is used to maintain the same proportion of classes in the train and test sets, it is important for imbalanced datasets like this one where the number of survivors is less than the number of non-survivors
```

```python
from sklearn.preprocessing import StandardScaler
sc= StandardScaler()
sc.fit(X_train[['Age','Fare']])
X_train[['Age','Fare']]= sc.transform(X_train[['Age','Fare']])
X_test[['Age','Fare']]= sc.transform(X_test[['Age','Fare']])

X_train
```

```python
y_train 
```

```python
from sklearn.linear_model import LogisticRegression
lr = LogisticRegression() #Logistic Regression is a linear model that is used for binary classification problems, it is used to predict the probability of the target variable being 1 or 0, it is a linear model because it assumes that there is a linear relationship between the features and the log odds of the target variable
lr.fit(X_train,y_train) #fit is used to train the model on the training data, it takes the features and the target variable as input and it learns the coefficients of the features that best fit the datat
y_pred= lr.predict(X_test) #predict is used to make predictions on the test data, it takes the features of the test data as input and it returns the predicted values of the target variable
y_pred
```

```python
#we use confusion matrix to evaluate the performance of the model, it is a table that is used to evaluate the performance of a classification model, it shows the number of true positives, true negatives, false positives and false negatives
from sklearn.metrics import confusion_matrix
cm = confusion_matrix(y_test,y_pred)
cm
```

```python
sns.heatmap(cm, annot=True, fmt='d') #annot = True is used to annotate the cells with the numbers, fmt = 'd' is used to format the numbers as integers
plt.xlabel('Predicted')
plt.ylabel('Actual')
plt.show()
```

```python
TN= 139
TP= 75
FP= 28
FN=26
```

```python
#Accuracy means the proportion of correct predictions out of the total predictions, it is calculated as (TP+TN)/(TP+TN+FP+FN)
#kitne sahi predict kiye aur kitne galat predict kare hain 
accuracy = (TP+TN)/(TP+TN+FP+FN)
print('Accuracy:', accuracy)
```

```python
import sklearn.metrics as accuracy_score
accuracy_score.accuracy_score(y_test,y_pred) #accuracy_score is used to calculate the accuracy of the model, it takes the true values and the predicted values as input and it returns the accuracy of the model
```

```python
#precison is the proportion of true positives out of the total predicted positives, it is calculated as TP/(TP+FP)
# sare positive predictions me se kitne sahi hain (focus on FP)
precision = TP/(TP+FP)
print('Precision',precision)
```

```python
import sklearn.metrics as precision_score
precision_score.precision_score(y_test,y_pred) #precision_score is used to calculate the precision of the model, it takes the true values and the predicted values as input and it returns the precision of the model
```

```python
#recall is the proportion of true positives out of the total actual positives, it is calculated as TP/(TP+FN)
# actually jitne positive the unme se kitne sahi predict kiye hain (focus on FN)
recall = TP/(TP+FN)
print('Recall',recall)
```

```python
import sklearn.metrics as recall_score
recall_score.recall_score(y_test,y_pred) #recall_score is used to calculate the recall of the model, it takes the true values and the predicted values as input and it returns the recall of the model
```

```python
# f1 score means the harmonic mean of precision and recall, it is calculated as 2*(precision*recall)/(precision + recall), it is used to evaluate the performance of a classification model when there is an imbalance in the classes, it gives more weight to the minority class
# f1 score matlab precision aur recall ka harmonic mean, isko calculate karne ke liye 2*(precision*recall)/(precision + recall) formula use hota hai, f1 score ko use kiya jata hai jab classes me imbalance hota hai, f1 score minority class ko zyada weight deta hai
f1 = 2*(precision*recall)/(precision + recall)
print('F1 Score:', f1) 
```

```python
import sklearn.metrics as f1_score
f1_score.f1_score(y_test,y_pred) #f1_score is used to calculate the f1 score of the model, it takes the true values and the predicted values as input and it returns the f1 score of the model
```

```python
# feature selection
features = ['GrLivArea','BedroomAbvGr']
x = hd[features]
y = hd['SalePrice']
```

```python
from sklearn.metrics import root_mean_squared_error
rmse = root_mean_squared_error(y_test, y_pred_lr)
rmse
```

```python
from sklearn.metrics import r2_score
r2_val = r2_score(y_test,y_pred_lr)
r2_val
```

```python
r2_val = r2_score(y_test,y_pred_poly)
r2_val
```

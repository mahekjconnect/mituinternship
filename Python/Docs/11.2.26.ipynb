# Cross Validation – Holdout Method

- Cross validation is used to evaluate how well a model generalizes to unseen data
- It helps detect overfitting and underfitting
- The holdout method is the simplest form of cross validation


# What is the Holdout Method?

- The dataset is split into two parts
- Training set → used to train the model
- Testing set → used to evaluate the model
- Common split ratios:
  - 70% train / 30% test
  - 80% train / 20% test


# Why Do We Use It?

- If we train and test on same data → model memorizes
- We want to test on unseen data
- Holdout simulates real-world prediction performance


# Steps in Holdout Method

- Split the dataset into training and testing sets
- Train the model on training data
- Make predictions on testing data
- Calculate evaluation metric (MAE, RMSE, R², etc.)
- Analyze performance


# Example (Conceptual)

- Total data: 1000 rows
- 800 rows → training
- 200 rows → testing
- Model learns from 800
- Model is evaluated on 200


# Advantages of Holdout Method

- Simple to implement
- Fast
- Works well for large datasets
- Easy to understand


# Disadvantages of Holdout Method

- Performance depends on how data is split
- Different splits may give different results
- Not reliable for small datasets
- May accidentally create biased splits


# Overfitting in Holdout

- If training accuracy is high
- But testing accuracy is low
- Model is overfitting


# When to Use Holdout Method

- When dataset is large
- When quick validation is needed
- When computational power is limited


# Key Idea

- Train on one part
- Test on unseen part
- Measure performance
- Check generalization ability

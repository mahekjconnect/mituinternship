There are two core objects in pandas: the DataFrames and the Series.

DataFrame:
A dataframe is a table, it contains an array of individual entries(column), each of which has a certain value(row), and these are indexed.


```python
import pandas as pd

df = pd.DataFrame({'mahek':[12,43,65,54], 'Sayma' :[23,45,67,89]})

print("DataFrame created successfully.")

print(df)
```

so in this example, mahek at '0' has value 12. and sayma at '0' has value 23.

we can also save strings in dataframes.

we are basically creating a dictionary whose keys are the column names and whose values are the list of entries.

we are using the pd.DataFrame() contructor to generate these DataFrame objects.

The list of row labels used in DataFrame is known as an Index.[0,1,2,3] we can assign values to it by using an index parametere in our contructor

```python
import pandas as pd
df= pd.DataFrame({'mahek':['wow!thats great!', 'umm, its ok ok'], 'sayma':['meh! i dont rly like it.', 'yayyy!! i love it!']},
                 index=['The park','The mall'])
print('Second DataFrame using strings and custom index:')
print(df)
print(df['mahek']['The park'])
```

so when we rename the index, its original name is changed, so basically now if u want to print 'the park' value of 'mahek'

loc and iloc are used for indexing and selecting specific rows and columns from pandas DataFrames and Series.

iloc - is used to do slicing using intergers [starting_row:ending_row, staring_column:ending_column]
Use when- 1. number of rows known 2. doing slicing of array or loops

loc - is used to do slicing using labels[staring_row_label:ending_row_labelstaring_col_label:ending_col_label]
Use when- 1.working with col names 2.flitering data 3.reable, production code

```python
df.iloc[2:5, 0:3]  #slicing using iloc[row_start:row_end, col_start:col_end]
```

```python
df.iloc[[5,6,7], 0:2]  #slicing using iloc to get specific rows and columns
```

```python
df.loc[3:6,['Name','Numeric']]  #slicing using loc[row_labels, column_labels]
```

```python
df['Abbreviation']  #accessing a single column
```

```python
df.drop(7)  #drops the row with index label 7
```

```python
df.drop('Numeric', axis=1)  #drops the column with label 'Numeric', and axis=1 indicates that we are dropping a column because columns are along axis 1
```

```python
col = {'Name': 'Month'}
df.rename(columns=col)  #renames the column 'Name' to 'Month'
```

```python
inx = {1:'one', 2:'two', 3:'three'}
df.rename(index=inx)  #renames the row labels 1,2,3 to one,two,three
```

```python
df['Numeric'] = df['Numeric'] * 2  #updates column 'Numeric' with double the values of column 'Numeric'
df
```

```python
df['Numeric-2']= df['Numeric-2'] +200  #updates 'Numeric-2' with values of column 'Numeric-2' plus 200
df
```

Series: a series is a sequence of data values, if a DataFrame is a teble, a series is a list. you can create a series with nothing more than a list.

```python
import pandas as pd
s = pd.Series([1,2,3,4,5])
print(s)
```

a series is a single column of dataframe, so we can assign row labels same as we did in dataframes, using the index parameter. however series does not have column name


```python
import pandas as pd
s = pd.Series(['happy','calm','bummer','cold'], index=['mahek','sayma', 'zia','khadu'])
print(s)
print(s['zia'])
```

soo basically, here we can use directly the index name and not specify any other thing kyuki bas ek hi column hai :)

# Overfitting and Underfitting

- These are common problems in Machine Learning models
- They affect how well a model performs on unseen data
- They are related to bias and variance


# What is Underfitting?

- Model is too simple
- Cannot capture the underlying pattern in data
- Performs poorly on training data
- Also performs poorly on testing data
- High bias
- Low variance

- Example:
  - Fitting a straight line to strongly curved data


# Signs of Underfitting

- High training error
- High testing error
- Model is not learning enough from data


# What is Overfitting?

- Model is too complex
- Learns noise and random fluctuations
- Performs very well on training data
- Performs poorly on testing data
- Low bias
- High variance

- Example:
  - High-degree polynomial passing through all points


# Signs of Overfitting

- Very low training error
- Very high testing error
- Model memorizes instead of generalizing


# Bias-Variance Relationship

- Bias = error due to overly simple model
- Variance = error due to model being too sensitive to training data

- Underfitting → High bias, Low variance
- Overfitting → Low bias, High variance


# Visual Intuition

- Underfitting → straight line on curved data
- Good fit → smooth curve following pattern
- Overfitting → wiggly curve touching all points


# Causes of Underfitting

- Model too simple
- Too few features
- Low model complexity
- Insufficient training time


# Causes of Overfitting

- Model too complex
- Too many features
- High polynomial degree
- Small dataset
- No regularization


# How to Fix Underfitting

- Increase model complexity
- Add more relevant features
- Increase polynomial degree
- Train longer
- Use a more powerful model


# How to Fix Overfitting

- Reduce model complexity
- Reduce polynomial degree
- Remove irrelevant features
- Use regularization (Ridge, Lasso)
- Use more training data
- Use cross-validation


# Key Idea

- Goal is to find balance between bias and variance
- Ideal model performs well on both training and testing data
- This balance is called the Bias-Variance Tradeoff

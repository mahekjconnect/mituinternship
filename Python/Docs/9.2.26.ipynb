# Linear Regression 

# What is Linear Regression?

- Linear Regression is used to predict continuous numerical values
- It finds the best straight line that fits the data
- It models the relationship between input (X) and output (Y)
- Example: predicting salary, house price, sales


# Equation of Linear Regression

- Basic straight line equation: y = mx + c
- In ML form: y = β0 + β1x
- β0 = intercept (starting value when x = 0)
- β1 = slope (how much y changes when x increases by 1)
- Positive slope → increasing relationship
- Negative slope → decreasing relationship


# How Linear Regression Finds the Best Line

- There are many possible lines
- Model chooses the line with least error
- Error = Actual value − Predicted value
- We square errors to remove negatives
- We take average of squared errors
- This is called Mean Squared Error (MSE)
- Best line = line with lowest MSE


# Assumptions of Linear Regression

- Linearity → relationship should be straight
- No multicollinearity → features should not be highly correlated
- Homoscedasticity → constant error variance
- Errors should be independent
- Errors should be normally distributed (mainly for statistical testing)


# Geometric Interpretation

- With 1 feature → model forms a straight line
- With 2 features → model forms a plane
- With many features → model forms a hyperplane
- It finds the best flat surface that fits data points


# Bias–Variance Concept

- Bias = error due to model being too simple
- High bias → underfitting
- Variance = error due to model being too complex
- High variance → overfitting
- Linear regression usually has low variance
- It can have high bias if relationship is nonlinear


# When to Use Linear Regression

- When relationship looks linear
- When interpretability is important
- When dataset is not extremely complex
- When you need a simple baseline model


# Advantages of Linear Regression

- Easy to understand
- Fast to train
- Works well on small datasets
- Easy to interpret coefficients
- Good starting model


# Disadvantages of Linear Regression

- Cannot model nonlinear relationships
- Sensitive to outliers
- Assumptions must be satisfied
- Multicollinearity can cause instability


# Real Life Applications

- House price prediction
- Salary prediction
- Sales forecasting
- Demand prediction
- Revenue estimation
- Basic trend analysis

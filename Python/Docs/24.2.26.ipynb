# Entropy

- Entropy measures how mixed or impure the data is
- It is mainly used in classification problems
- High entropy → Data is very mixed
- Low entropy → Data is more pure
- Entropy value ranges between 0 and 1

- If all data belongs to one class → Entropy = 0 (pure)
- If data is equally mixed → Entropy is maximum

# Information Gain

- Information Gain tells how much entropy decreases after a split
- Decision Tree chooses the split with highest Information Gain
- Higher Information Gain → Better split

# Gini Index

- Gini Index also measures impurity of data
- Used in classification problems
- Gini value ranges between 0 and 1
- Gini = 0 → Perfectly pure node
- Higher Gini → More mixed data

# Difference Between Entropy and Gini

- Both measure impurity
- Both are used to decide best split
- Entropy uses logarithm
- Gini is computationally faster
- In practice, results are very similar

# Pruning

- Pruning means cutting unnecessary branches of the tree
- It helps reduce overfitting
- It makes the model simpler and better for new data

# Why Pruning is Needed

- Decision Trees tend to grow very deep
- Deep trees memorize training data
- This causes overfitting

# Types of Pruning

- Pre-Pruning
  - Stop tree growth early
  - Use max_depth
  - Use min_samples_split
  - Use min_samples_leaf

- Post-Pruning
  - Grow full tree first
  - Remove weak branches later

# Summary

- Entropy → Measures disorder
- Gini Index → Measures impurity
- Information Gain → Reduction in entropy after split
- Pruning → Prevents overfitting by simplifying tree

# Feature Extraction

- Process of creating new features from existing data
- Transforms data into better representation

# Need for Feature Extraction

- Reduces dimensionality (number of features)
- Improves model performance
- Faster training
- Removes noise
- Helps in data visualization

# Feature Selection vs Feature Extraction

- Feature Selection → selects important existing features
- Feature Extraction → creates new features from old ones


# Techniques of Feature Extraction

# 1. PCA (Principal Component Analysis)

- Converts original features into new features (principal components)
- First component captures maximum variance
- Components are uncorrelated
- Used for dimensionality reduction


# 2. LDA (Linear Discriminant Analysis)

- Used for classification problems
- Maximizes separation between classes
- Minimizes variation within class


# 3. Image Feature Extraction (Basic)

- Converts images into numerical features
- Extracts edges, shapes, textures
- Used in computer vision tasks


# Key Difference

- Selection → choose features
- Extraction → create features

# Feature Extraction in NLP

- Text data cannot be used directly in ML models
- We convert text into numerical features
- This process is called feature extraction in NLP


# Why needed

- ML models understand only numbers
- Converts words/sentences into vectors

# Feature Selection

Feature Selection = choosing important features (columns) from dataset
Removes unnecessary, irrelevant, and redundant features


Why we need it:

- Improves model accuracy
- Reduces overfitting
- Makes training faster
- Reduces noise
- Makes model simple & easy to understand


Types of Feature Selection:

- Filter Methods
- Wrapper Methods
- Embedded Methods


Filter Methods:

- Use statistical techniques
- Do NOT use ML model
- Check only relevance of features
- Fast and simple

Techniques:

Information Gain (IG):
- Measures how much a feature helps predict target
- Higher value → more important

Chi-Square Test (χ²):
- Used for categorical data
- Checks dependency between feature & target
- More dependency → better feature

Correlation Coefficient:
- Measures linear relationship (-1 to +1)
- Remove highly correlated features (avoid redundancy)


Wrapper Methods:

- Uses ML model to evaluate features
- Trains model multiple times
- Checks usefulness of features
- More accurate but slow

Techniques:

RFE (Recursive Feature Elimination):
- Start with all features
- Remove least important step-by-step
- Get best subset

Genetic Algorithm:
- Based on natural selection
- Tries different feature combinations
- Keeps best performing ones


Embedded Methods:

- Feature selection happens during training
- Combines filter + wrapper benefits
- Checks both relevance and usefulness

Techniques:

Decision Trees:
- Automatically select important features
- Uses IG / Gini Index
- Important features appear at top


Final Idea:

Filter → Fast, checks relevance  
Wrapper → Slow, checks usefulness  
Embedded → Balanced, checks both
